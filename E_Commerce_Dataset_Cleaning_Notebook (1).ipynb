# **About Dataset**
## This dataset contains 5 different CSV files representing distinct categories of e-commerce data. Below is a comprehensive Python notebook detailing the data cleaning pipiline using Google Colab and Pandas.

```python
import pandas as pd
from google.colab import files
```

# **1.Cleaning Process for Women clothing data(CSV)**

```python
# loading csv file of womens_clothing
path='/content/drive/MyDrive/e-commerce dataset/us-shein-womens_clothing-4620.csv'
data=pd.read_csv(path,encoding='latin-1')
data.head()
```

```python
data.info()# initial data information
```

```python
# let's first rename the column name to look professional
df_cleaned=data.rename(columns={'goods-title-link':'product_name',
                                'color-count':'available_color',
                                'selling_proposition':'tagline',
                                'product-locatelabels-img src':'image_url'}).copy()
df_cleaned.info()
```

```python
# Now change the data type of price and discount for analysis
if 'price' in df_cleaned.columns:
  df_cleaned['price']=df_cleaned['price'].astype(str).str.replace(r'[^\d.]','',regex=True)
  df_cleaned['price']=pd.to_numeric(df_cleaned['price'],errors='coerce')
```

```python
# Now change the data type of discount
if 'discount' in df_cleaned.columns:
  df_cleaned['discount']=df_cleaned['discount'].astype(str).str.replace(r'[^\d.]','',regex=True)
  df_cleaned['discount']=pd.to_numeric(df_cleaned['discount'],errors='coerce')
```

```python
# Now fill missing discount with 0 means no discount
df_cleaned['discount']=df_cleaned['discount'].fillna(0)
```

```python
# Now fill missing available_color with 1 means assuming at least the default color
df_cleaned['available_color']=df_cleaned['available_color'].fillna(1)
```

```python
# Now add our trekking categoris for the master dataset later
df_cleaned['main_category']="Women's Clothing"
```

```python
# Now let's check
df_cleaned.info()
```

```python
df_cleaned.head()
```

```python
df_cleaned.to_csv('clean_womens_clothing_data.csv',index=False)
files.download('clean_womens_clothing_data.csv')
```

# **3. Cleaning Proceses for Baby and Maternity data(CSV)**

```python
path='/content/drive/MyDrive/e-commerce dataset/us-shein-baby_and_maternity-4433.csv'
data=pd.read_csv(path,encoding='latin-1')
data.head()
```

```python
data.info() # before cleaning
```

```python
# first merge column goods title link jump and goods title link as product title
data['product_title']=data['goods-title-link--jump'].fillna(data['goods-title-link'])
data.head()
```

```python
# now drop the useless columns
columns_to_drop=['goods-title-link--jump','goods-title-link']
data=data.drop(columns=[col for col in columns_to_drop if col in data.columns])
data.info()
```

```python
# Now change the columns name for analysis
df_cleaned=data.rename(columns={'color-count':'available_color',
                                'goods-title-link--jump href':'title_href'}).copy()
df_cleaned.info()
```

```python
#Now change the data type of available color as float to numeric
if 'available_color' in df_cleaned.columns:
  df_cleaned['available_color']=pd.to_numeric(df_cleaned['available_color'])
```

```python
# Now change the data type of price and discount for analysis
if 'price' in df_cleaned.columns:
  df_cleaned['price']=df_cleaned['price'].astype(str).str.replace(r'[^\d.]','',regex=True)
  df_cleaned['price']=pd.to_numeric(df_cleaned['price'],errors='coerce')
# Now change the data type of discount
if 'discount' in df_cleaned.columns:
  df_cleaned['discount']=df_cleaned['discount'].astype(str).str.replace(r'[^\d.]','',regex=True)
  df_cleaned['discount']=pd.to_numeric(df_cleaned['discount'],errors='coerce')

# Now fill missing discount with 0 means no discount
df_cleaned['discount']=df_cleaned['discount'].fillna(0)

# Now fill missing available_color with 1 means assuming at least the default color
df_cleaned['available_color']=df_cleaned['available_color'].fillna(1)

# Now add our trekking categoris for the master dataset later
df_cleaned['main_category']="Baby & Maternity"

```

```python
df_cleaned.head()
```

```python
df_cleaned.info()
```
